```python
flights["on_time_percent"] = (
    flights["flights_15_minutes_early_to_1_minute_early_percent"]
    + flights["flights_0_to_15_minutes_late_percent"]
)
```

```python
flights[
    [
        "flights_15_minutes_early_to_1_minute_early_percent",
        "flights_0_to_15_minutes_late_percent",
        "on_time_percent"
    ]
].head()
```

```python
flights["delayed_over_15_percent"] = (
    flights["flights_between_16_and_30_minutes_late_percent"]
    + flights["flights_between_31_and_60_minutes_late_percent"]
    + flights["flights_between_61_and_120_minutes_late_percent"]
    + flights["flights_between_121_and_180_minutes_late_percent"]
    + flights["flights_between_181_and_360_minutes_late_percent"]
    + flights["flights_more_than_360_minutes_late_percent"]
)
```

```python
flights["severe_delay_percent"] = (
    flights["flights_between_61_and_120_minutes_late_percent"]
    + flights["flights_between_121_and_180_minutes_late_percent"]
    + flights["flights_between_181_and_360_minutes_late_percent"]
    + flights["flights_more_than_360_minutes_late_percent"]
)
```

```python
flights[
    [
        "on_time_percent",
        "delayed_over_15_percent",
        "severe_delay_percent",
        "average_delay_mins"
    ]
].head(10)
```

```python
flights["estimated_delayed_flights"] = (
    flights["number_flights_matched"]
    * flights["delayed_over_15_percent"]
    / 100
)
```

```python
flights["estimated_severe_delay_flights"] = (
    flights["number_flights_matched"]
    * flights["severe_delay_percent"]
    / 100
)
```

```python
flights[
    [
        "number_flights_matched",
        "delayed_over_15_percent",
        "estimated_delayed_flights",
        "severe_delay_percent",
        "estimated_severe_delay_flights"
    ]
].head(10)
```
